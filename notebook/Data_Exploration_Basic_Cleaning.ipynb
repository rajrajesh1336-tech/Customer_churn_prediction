```python
#imports
import numpy as np
import pandas as pd
```

```python
# load Data
data = pd.read_csv(r"D:\ML_Project\Customer_churn_predection\Data\Telco_Customer_Churn.csv")
data.head()
```

```python
# Backup data and info
df = data.copy()
df.info()
```

```python
# Meta-Data
print("No. of rows:", df.shape[0])
print("No. of Columns:", df.shape[1])
all_Columns = df.columns.to_list()
num_Columns = df.select_dtypes(include = "number").columns.to_list()
cat_Columns = df.select_dtypes(include = "object").columns.to_list()

print("All Columns: \n", all_Columns)
print("\nNumeric Columns: \n", num_Columns)
print("\nCategorical Columns: \n", cat_Columns)

print("\nData Types: \n", df.dtypes)
```

```python
# Duplicate
print("Duplicated count: - ", df.duplicated().sum())
print("Duplicate Rows")
df[df.duplicated()]
```

```python
#Drop duplicates
df.drop_duplicates()
print("Duplicated count after deleation", df.duplicated().sum())
```

```python
# Remove irrelevant column "customerID"
df.drop(columns = "customerID", inplace = True)
df.head(2)
```

```python
# Change Data-Type of "TotalCharges"
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors = 'coerce')
df.dtypes
```

```python
# Save clean Data
clean_data_path = r"D:\ML_Project\Customer_churn_predection\Data\cleaned_data.csv"
df.to_csv(clean_data_path, index = False)
print("Cleaned data is save as cleaned_data in D:\ML_Project\Customer_churn_predection\Data")
```
